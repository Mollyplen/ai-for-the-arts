## LLM - Music Heritage <br>

My design is for an application that uses a large language model to support music heritage. The idea is to design a tool that helps people understand music in context, much like exhibition labels or museum interpretation panels, but in a format that can adapt to different audiences and needs. <br>

This idea comes from my own experiences this year visiting music museums like The British Music Experience, the Irish Rock ’n’ Roll Museum Experience, and The Beatles Museum. I already love learning about music, but seeing the history and stories around an artist or song in a curated space genuinely changes how you listen. Once you understand the wider picture, the music takes on a new meaning, and you end up with a different kind of appreciation. It’s hard to hear it the same way again after that.<br>

That made me think it would be exciting to design an LLM-based tool that can recreate some of that “museum context” feeling for people who can’t easily access these experiences. Not as a replacement for museums, but as a way to make music heritage learning more accessible, so you can listen to a song or album and learn at the same time, without needing to be in a physical curated space. Not everyone can travel, afford tickets, or feel confident in heritage spaces, so a tool like this could lower the barrier to engaging with music history. 
This idea could work for different ages, learning styles, and language backgrounds. A lot of heritage spaces are moving away from “just stand and read a label on the wall” and becoming more immersive and interactive, which is great, but barriers still exist. Some people need information in shorter chunks, some prefer an audio-style explanation, and some feel overwhelmed. On top of that, often resources at museums and heritage sites assume strong English, which can make the stories harder to access for visitors with different first languages. A tool like this could adapt the same music heritage context with different formats and levels of complexity, so children or casual listeners can also engage with it in a way that suits them. <br>

This design could work both in museums and anywhere else. In a museum, it could act like a QR companion where you scan an object or label and instantly get the story in the right format (short, kid-friendly, audio-style, or translated). Outside museums, it could be used while listening at home or on the go, and could even be built into apps like Spotify as an optional “context” tab beside the track.<br>

### What Application Would Do <br>
The application would take an input from a user, such as an album, song, or interview extract, and produce a few different outputs depending on what the user chooses. The main options would be… 
- Explain in plain language - a simple explanation of what the text is saying, without assuming prior knowledge. 
- Liner notes style summary - A short, engaging paragraph that adds context and highlights why it matters. 
- Glossary mode - A list of key terms and references explained clearly, such as genre terms, instruments, people, and historical events. 
- Timeline mode - A short timeline of what was happening around this music to show historical and cultural context.
- Disscussion - questions for learning or group discussions that encourage interpretation not just facts. 
- Accessibility features - Settings like shorter sentences, reading level, bullet version, transations, and audio wording.

### How Users would Interact<br>
**Example Interaction:**
1. User: “I’m listening to this album, I want to learn about the context behind it, but I don't want an essay” 
2. AI: Tell me the album and what kind of output you want. Who is the audience (age/background), and would you like a glossary? 
3. User: Audience is a general listener, keep it simple to understand. I also want a glossary. 
**AI Returns:**
- 1 short liner notes paragraph
- 5 glossary terms
- 3 questions to think about while listening 
4. User: “Can you make it more suitable for someone younger and keep sentences short?”
AI: Rewrites in shorter sentences, removes technical jargon, and keeps the same main meaning. 

### How LLM Works <br>
The user gives an input, such as a song title, album, artist, or even an interview clip. The system then turns that input into a format the model can work with. If the input is text, it is already “language data”. If the input is a song, the app would first identify it, either by user typing it, choosing from streaming link, or using a shazam-style “listen” tool that recognises the track. The app would then pull basic metadata such as the artist name, album and genre. <br>

The LLM itself doesnt “know” music history in a way a human does. It predicts likely words based on patterns it learned during training. In this design, the model is used as an interpretation tool, it takes the retrieved context and transforms it into different outputs depending on the users chosen mode and accessibility features. This is human and ai interaction rather than ai working alone, the person stays in controlby choosing the output style, reading level, language, and what they want to focus on. <br>

### Design Choices to reduce mistakes <br>
Because LLM’s can sound confident even when they are wrong, the app would be designed in a way to reduce “hallucinations” and misrepresentation, especially since heritage is sensitive and can involve identity, politics, and contested narratives. <br>

The app would only pull from a set of trusted sources ,such as museum pages, archives, recogised biographies, and use that as the base for the response, rather than letting the model make things up from memory. <br>

The souce would be clearly displayed so users could see citations or links to build trust with the app.<br>

If the model could’t confirm something it should say if it is unclear instead of guessing. 
For younger users it should avoid mature content and keep explanations appropriate. 
The app would not reproduce full lyrics due to copyright, it could summarise and interpret instead, and also link out to other sources. 

### Critical Reflection 
The design could make music heritage learning more inclusive, but it also comes with risks. A major benefit is its accessibility as it can help those cant travel to or access heritage spaces. It supports different learning styles and interactive questions rather than one fixed “museum label” format. For language access, translation features could help visitors who are often excluded. <br>

However, there are cultural risks. If the LLM is trained on biased data, it may repeat them or ignore unrepresented voices. I tool like this could accidentally turn complicated histories into simple “fun facts” and present misinformation. Museums have a responsibilities around ethical and representation, and an app could undermine that if it replaces lived experiences with automated responses.
